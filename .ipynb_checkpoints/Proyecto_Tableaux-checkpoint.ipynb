## Reglas <a class="anchor" id="forms"></a>

Para hacer más concisa la primera regla, hacemos la siguiente definición de lo que son las lineas rectas en el hexagrama (tenga en cuenta que las lineas son un subconjunto de todas las casillas):

$$l_1 = \{1,3,6,8\}$$
$$l_2 = \{1,4,7,11\}$$
$$l_3 = \{8,9,10,11\}$$
$$l_4 = \{2,3,4,5\}$$
$$l_5 = \{2,6,9,12\}$$
$$l_6 = \{5,7,10,12\}$$

Las restricciones y su representación usando lógica proposicional son las siguientes:

1. La suma de los números en cualquier línea recta es 26:

$$\bigwedge_{l \in Lineas} \left(\bigvee_{n \in Numeros} NenC_{n,l_{[1]}}+NenC_{n,l_{[2]}}+NenC_{n,l_{[3]}}+NenC_{n,l_{[4]}} = NenC_{26,l_{[1]}+l_{[2]}+l_{[3]}+l_{[4]}}\right)$$

2. Todas las casillas deben estar ocupadas:
$$\bigwedge_{c \in Casillas} \left(\bigvee_{n \in Numeros}NenC_{n,c}\right)$$


3. No pueden haber números repetidos:
$$\bigwedge_{n \in Numeros} \bigwedge_{c \in Casillas}\left(NenC_{n,c} \rightarrow \neg \left(\bigvee_{e \neq c} NenC_{n,e}\right)\right)$$


4. Solo debe haber un número por casilla:
$$\bigwedge_{n \in Numeros} \bigwedge_{c \in Casillas}\left(NenC_{n,c} \rightarrow \neg \left(\bigvee_{m \neq n} NenC_{m,c}\right)\right)$$

## Expansión primero en anchura <a class="anchor" id="anchura"></a>

<img src="./img/anchura.png" width="350px"/>

La implementación en Python es la siguiente:

```python
def primero_anchura(nodo):
    estado = nodo
    res = estado.es_hoja()
    if res == 'cerrada':
        return None
    elif res == 'abierta':
        return estado.interp()
    frontera = [estado]
    while len(frontera) > 0:
        estado = frontera.pop(0) 
        hijos = estado.expandir()
        for a in hijos:
            res = a.es_hoja()
            if res == 'abierta':
                return a.interp()
            elif res == None:
                frontera.append(a)
    return None
```

### Regla 1

```python
%%time
A = [inorder_to_tree(n.reglas[0])]
n1 = nodos_tableaux(A)
I1 = primero_anchura(n1)
print("Resultado:", I1)
```

```python
%%time
if I1 != None:
    n.visualizar(I1)
else:
    print('¡No hay solución!')
```

# Expansión primero en profundidad <a class="anchor" id="prof"></a>

([Volver al inicio](#inicio))

Además de poder expandir primero en anchura, podemos expandir primero en profundidad. El pseudocódigo de este algoritmo es el siguiente:

<img src="./img/profundidad.png" width="350px"/>

```python
def primero_profundidad(nodo):
    estado = nodo
    res = estado.es_hoja()
    if res == 'cerrada':
        return None
    elif res == 'abierta':
        return estado.interp()
    frontera = [estado]
    while len(frontera) > 0:
        estado = frontera.pop() 
        hijos = estado.expandir()
        for a in hijos:
            res = a.es_hoja()
            if res == 'abierta':
                return a.interp()
            elif res == None:
                frontera.append(a)
    return None
```

# Recorrido con backtracking <a class="anchor" id="backt"></a>

([Volver al inicio](#inicio))

Podemos hacer también un recorrido por el árbol haciendo una expansión recursiva que implementa el backtracking, como en el siguiente pseudocódigo:

<img src="./img/backtracking.png" width="350px"/>

```python
def backtracking(nodo):
    estado = nodo
    res = estado.es_hoja()
    if res == 'cerrada':
        return None
    elif res == 'abierta':
        return estado.interp()
    hijos = estado.expandir()
    for hijo in hijos:
        res = backtracking(hijo)
        if res is not None:
            return res
    return None
```

```python
forms3 = ['--(-(pOq)Y-(r>s))']
forms3 = [inorder_to_tree(f) for f in forms3]
m3 = nodos_tableaux(forms3)
I3 = backtracking(m3)
print("Resultado:", I3)
```

```python
forms4 = ['(((pYq)O(rOq))Y(-qO-p))']
forms4 = [inorder_to_tree(f) for f in forms4]
m4 = nodos_tableaux(forms4)
I4 = backtracking(m4)
print("Resultado:", I4)
```
