## Reglas <a class="anchor" id="forms"></a>

Para hacer más concisa la primera regla, hacemos la siguiente definición de lo que son las lineas rectas en el hexagrama (tenga en cuenta que las lineas son un subconjunto de todas las casillas):

$$l_1 = \{1,3,6,8\}$$
$$l_2 = \{1,4,7,11\}$$
$$l_3 = \{8,9,10,11\}$$
$$l_4 = \{2,3,4,5\}$$
$$l_5 = \{2,6,9,12\}$$
$$l_6 = \{5,7,10,12\}$$

Las restricciones y su representación usando lógica proposicional son las siguientes:

1. La suma de los números en cualquier línea recta es 26:

$$\bigwedge_{l \in Lineas} \left(\bigvee_{n \in Numeros} NenC_{n,l_{[1]}}+NenC_{n,l_{[2]}}+NenC_{n,l_{[3]}}+NenC_{n,l_{[4]}} = NenC_{26,l_{[1]}+l_{[2]}+l_{[3]}+l_{[4]}}\right)$$

2. Todas las casillas deben estar ocupadas:
$$\bigwedge_{c \in Casillas} \left(\bigvee_{n \in Numeros}NenC_{n,c}\right)$$


3. No pueden haber números repetidos:
$$\bigwedge_{n \in Numeros} \bigwedge_{c \in Casillas}\left(NenC_{n,c} \rightarrow \neg \left(\bigvee_{e \neq c} NenC_{n,e}\right)\right)$$


4. Solo debe haber un número por casilla:
$$\bigwedge_{n \in Numeros} \bigwedge_{c \in Casillas}\left(NenC_{n,c} \rightarrow \neg \left(\bigvee_{m \neq n} NenC_{m,c}\right)\right)$$

```python
import pycosat

def MiniSAT(A):
    
    def lit_numero(l):
        if '-' in l:
            return -(ord(l[1:]) - 255)
        else:
            return ord(l) - 255
    
    def clausula_numero(C):
        return [lit_numero(l) for l in C]

    def fnc_numero(S):
        return [clausula_numero(C) for C in S]

    def obtener_int(mod):
        print("obtener_int")
        return {chr(255 + abs(n)):n>0 for n in mod}
    
    def obtener_numeros(S):
        lista_plana = [abs(x) for sublist in S for x in sublist]
        lista_plana = list(set(lista_plana))
        return lista_plana
        
    S = tseitin(A)
    print(S)
    S = fnc_numero(S)
    print(S)
    numeros = obtener_numeros(S)
    print(numeros)
    solucion = pycosat.solve(S)
    solucion = [x for x in solucion if abs(x) in numeros]
    print(solucion)
    solucion = obtener_int(solucion)
    print(solucion)
    with Minisat22(bootstrap_with=S) as m:
        if m.solve():
            return 'Satisfacible', obtener_int(m.get_model())
        else:
            return 'Insatisfacible', {}
```

```python
def filtro(I, formula_original):
    """
    Filtra las variables de Tseitin, dejando solo las originales.
    Maneja casos donde la interpretación es None (Insatisfacible).
    """
    if I is None or I == {}:
        return I

    # Convertimos a set para búsquedas instantáneas (O(1))
    # Esto es vital si tienes cientos de variables de Tseitin
    letras_reales = set(inorder_to_tree(formula_original).letras())
    
    # Construimos el nuevo diccionario filtrado
    return {var: val for var, val in I.items() if var in letras_reales}

    #Uso:
    #I_final = filtro(I_res, F)
```
