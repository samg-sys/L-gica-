$$l_1 = \{1,3,6,8\}$$
$$l_2 = \{1,4,7,11\}$$

$$l_4 = \{2,3,4,5\}$$

Las restricciones y su representación usando lógica proposicional son las siguientes:

1. La suma de los números en cualquier línea recta es 26:

$$\bigwedge_{l \in Lineas} \left(\bigvee_{n \in Numeros} NenC_{n,l_{[1]}}+NenC_{n,l_{[2]}}+NenC_{n,l_{[3]}}+NenC_{n,l_{[4]}} = NenC_{26,l_{[1]}+l_{[2]}+l_{[3]}+l_{[4]}}\right)$$

2. Todas las casillas deben estar ocupadas:
$$\bigwedge_{c \in Casillas} \left(\bigvee_{n \in Numeros}NenC_{n,c}\right)$$


3. No pueden haber números repetidos:
$$\bigwedge_{n \in Numeros} \bigwedge_{c \in Casillas}\left(NenC_{n,c} \rightarrow \neg \left(\bigvee_{e \neq c} NenC_{n,e}\right)\right)$$


4. Solo debe haber un número por casilla:
$$\bigwedge_{n \in Numeros} \bigwedge_{c \in Casillas}\left(NenC_{n,c} \rightarrow \neg \left(\bigvee_{m \neq n} NenC_{m,c}\right)\right)$$

# Importar funciones para DPLL

```python
def complemento(L):
    return L[1:] if L.startswith('-') else '-' + L
```

```python
def eliminar_literal(S, l):
    # S1 keeps only clauses where l is NOT present (because those clauses are satisfied)
    S1 = [c for c in S if l not in c]
    lc = complemento(l)
    # Then, remove the complement lc from the remaining clauses
    return [[p for p in c if p != lc] for c in S1]

def extender_I(I, L):
    # Extract variable name (remove '-' if present)
    var = L[1:] if L.startswith('-') else L
    # Value is True if no '-', False if '-'
    valor = not L.startswith('-')
    I[var] = valor
    return I
```

```python
def unit_propagate(S, I):
    cambio = True
    while cambio:
        cambio = False
        for c in S:
            if len(c) == 1:  # Found a unit clause
                L = c[0]
                var = L[1:] if L.startswith('-') else L
                val = not L.startswith('-')
                
                # Check for contradiction before proceeding
                if var in I and I[var] != val:
                    return [[]], I # Return a set containing the empty clause to trigger 'Insatisfacible'
                
                I = extender_I(I, L)
                S = eliminar_literal(S, L)
                cambio = True
                break
    return S, I



def unit_propagate_optimized(S, I):
    """
    Improved unit propagation that detects contradictions earlier.
    """
    S = [c[:] for c in S] # Shallow copy of clauses
    changed = True
    while changed:
        changed = False
        units = [c[0] for c in S if len(c) == 1]
        for l in units:
            var = l[1:] if l.startswith('-') else l
            val = not l.startswith('-')
            
            if var in I:
                if I[var] != val: return S, I, True # Contradiction found
                continue
            
            I[var] = val
            S = eliminar_literal(S, l)
            changed = True
            if not S: break
    return S, I, False
```

## DPLL

```python
def dpll(S, I): #dpll optimized
    """
    An optimized version of DPLL that reduces memory overhead 
    and handles unit propagation more robustly.
    """
    # 1. Unit Propagation (Modifies I and S locally)
    S, I, conflict = unit_propagate_optimized(S, I)
    if conflict:
        return "Insatisfacible", {}

    # 2. Base Cases
    if len(S) == 0:
        return "Satisfacible", I
    if any(len(c) == 0 for c in S):
        return "Insatisfacible", {}

    # 3. Branching (Heuristic: pick a literal from the shortest clause)
    # This is more efficient than purely random choice
    S = sorted(S, key=len)
    l = S[0][0]
    lc = complemento(l)

    # 4. Recursive Step
    # We only deepcopy when branching to explore the first path
    # Path 1: Try literal l as True
    res, new_I = dpll(eliminar_literal(S, l), extender_I(deepcopy(I), l))
    if res == "Satisfacible":
        return res, new_I

    # Path 2: Try literal l as False (its complement lc as True)
    return dpll(eliminar_literal(S, lc), extender_I(deepcopy(I), lc))
```

**Filtrado de letras**

```python
def filtro(I, formula_original):
    """
    Filtra las variables de Tseitin, dejando solo las originales.
    Maneja casos donde la interpretación es None (Insatisfacible).
    """
    if I is None or I == {}:
        return I

    # Convertimos a set para búsquedas instantáneas (O(1))
    # Esto es vital si tienes cientos de variables de Tseitin
    letras_reales = set(inorder_to_tree(formula_original).letras())
    
    # Construimos el nuevo diccionario filtrado
    return {var: val for var, val in I.items() if var in letras_reales}

    # Uso:
    #I_final = filtro(I_res, F)
```

```python
def f_eval(formula, I):
    # Base case: it's a single letter/atom
    if type(formula) == str:
        return I.get(formula, False)
    
    # Logic for Negation
    if formula.label == '-':
        return not f_eval(formula.right, I)
    
    # Logic for AND (Y)
    if formula.label == 'Y':
        return f_eval(formula.left, I) and f_eval(formula.right, I)
    
    # Logic for OR (O)
    if formula.label == 'O':
        return f_eval(formula.left, I) or f_eval(formula.right, I)
    
    # Logic for Implication (>)
    if formula.label == '>':
        return not f_eval(formula.left, I) or f_eval(formula.right, I)
```

```python
def verificar_modelo(reglas, I):
    for i, r in enumerate(reglas):
        # 1. Convertimos el string a un árbol lógico primero
        arbol = inorder_to_tree(r) 
        # 2. Usamos el método nativo de la librería para evaluarlo
        if not arbol.valor(I):     
            print(f"FALLO: La regla {i} no se cumple.")
            return False
    print("ÉXITO: El modelo es 100% correcto.")
    return True
```
